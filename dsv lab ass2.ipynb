## Question 30
Read a copied table into lab_df and perform all the requested column operations in sequence.

```python
lab_df = pd.read_clipboard()
print(lab_df)
print(lab_df.columns)
print(lab_df.iloc[:, 0])
space_columns = [col for col in lab_df.columns if ' ' in str(col)]
if space_columns:
    print(lab_df[space_columns[0]])
small_lab_df = lab_df.iloc[:, :3]
print(small_lab_df)
lab_df['New Column'] = 'Done'
print(lab_df)
lab_df = lab_df.drop(columns=['New Column'])
lab_df
```
